## define the funcation two return two argument 
Imagine you are writing a program for a shop keeper. A customer comes in with a total amount of money and wants to buy as many candies as possible. Each candy costs $5.We want to write a function that calculates two things at once:How many candies they can buy.How much change (money) they have left over.

```python
def calCandy(amount):

    quantity =amount//5
    remain=amount%5

    return quantity,remain

##calCandy(10)

calCandy(12)
```
